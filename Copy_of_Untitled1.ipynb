```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix
from matplotlib.colors import ListedColormap
from sklearn.metrics import confusion_matrix,accuracy_score
from sklearn import model_selection
from sklearn.ensemble import BaggingClassifier
from sklearn.naive_bayes import GaussianNB
import warnings
warnings.filterwarnings('ignore')

dataset = pd.read_csv('/iris.csv')
dataset.head()
```

```python
x=dataset.iloc[:,[0,1,2]]
x
```

```python
y=dataset.iloc[:,[4]]
y
```

```python
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=0)

Single = GaussianNB()
Single.fit(X_train,y_train)
```

```python
y_pred=Single.predict(X_test)
print("predicted values for single Naive Bayes classifier",y_pred)

```

```python
Acc_single = accuracy_score(y_test,y_pred)*100
print("Accuracy of single Naive Bayes classifier",Acc_single)
```

```python
cm = confusion_matrix(y_test,y_pred)
print("\n\n Confusion Matrix-using single Naive Bayes Classifier: \n",cm)
```

```python
base_cls = GaussianNB()
num_class = 100
```

```python
Bag = BaggingClassifier(estimator=base_cls,n_estimators=num_class,random_state=0)
Bag.fit(X_train,y_train)
```

```python
results = model_selection.cross_val_score(Bag,X_test,y_test,cv=10)
print("\n\nAccuracy using Bagged Set of Naive Bayes Classifiers:",results.mean()*100)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix
from matplotlib.colors import ListedColormap
from sklearn.metrics import confusion_matrix,accuracy_score
from sklearn import model_selection
from sklearn.ensemble import AdaBoostClassifier

import warnings
warnings.filterwarnings('ignore')

dataset = pd.read_csv('/iris.csv')
dataset.head()
```

```python
x=dataset.iloc[:,[0,1,2]]
x
```

```python
y=dataset.iloc[:,[4]]
y
```

```python
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=0)

sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)

adaboost = AdaBoostClassifier(n_estimators=100)
adaboost.fit(X_train,y_train)
```

```python
adaboost.score(X_test,y_test)
```

```python
y_pred = adaboost.predict(X_test)
print("predicted values for AdaBoostClassifier:")
y_pred
```

```python
Acc_adaboost = accuracy_score(y_test,y_pred)*100
print("\n\n Test Accuracy using AdaBoostClassifier:",Acc_adaboost)
```

```python
cm = confusion_matrix(y_test,y_pred)
print("\n\n Confusion Matrix-using AdaBoostClassifier: \n",cm)
```

```python
fig, ax = plt.subplots(figsize=(6,6))
ax.imshow(cm)
ax.grid(False)
ax.xaxis.set(ticks=(0,1,2), ticklabels=('Predicted 0', 'Predicted 1', 'Predicted 2'))
ax.yaxis.set(ticks=(0,1,2), ticklabels=('Actual 0', 'Actual 1', 'Actual 2'))
ax.set_ylim(2.5, -0.5)
for i in range(3):
    for j in range(3):
        ax.text(j, i, cm[i, j], ha='center', va='center', color='red')
plt.show()
```
